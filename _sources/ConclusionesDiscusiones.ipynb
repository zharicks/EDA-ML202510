# **Conclusiones y discusiones**

## **Conclusiones**

Los resultados obtenidos a lo largo de este apartado evidencian que la combinación de un preprocesamiento cuidadoso, la validación cruzada estratificada y un enfoque de balanceo de clases (SMOTE) permite construir modelos con alta capacidad de discriminación en un contexto multiclase marcado por el desbalance.

La comparación global de once métodos reveló que XGBoost (clásico y optimizado) y Random Forest constituyen los enfoques más robustos, alcanzando los mejores valores de F1-score (macro) y AUC (macro). Por su parte, KNN y las variantes de SVM (con kernels lineales y aproximaciones RBF) ofrecieron un rendimiento satisfactorio, aunque en general pordebajo de los ensambles y métodos de boosting. Modelos más simples, como Naive Bayes o Regresión Logística, encontraron limitaciones para capturar la complejidad intrínseca del fenómeno, en especial la detección de desenlaces graves (hospitalización o muerte).

Adicionalmente, se amplió la comparativa incluyendo métodos avanzados como Balanced Random Forest, StackingClassifier, SMOTEBoost y SMOTE + AdaBoost (RandomizedSearchCV). Para profundizar en la identificación de casos de riesgo, se redefinió la variable objetivo hacia una clasificación binaria (`Victima_Peligro`) y se repitió el entrenamiento con SMOTE sobre cada fold. Tras esta transformación y el reentrenamiento de XGBoost, Random Forest y los nuevos modelos, ninguno logró superar el F1-score macro alcanzado originalmente por XGBoost clásico y Random Forest.

En el conjunto de modelos evaluados, XGBoost clásico sobresale al equilibrar un excelente poder predictivo con un tiempo de entrenamiento razonable, mientras que la versión optimizada reduce el tiempo de cómputo sin sacrificar rendimiento. Random Forest, por su parte, se presenta como una alternativa sólida y más interpretable, lo que podría facilitar su adopción en
ámbitos de salud pública.

Estos hallazgos confirman la pertinencia de combinar técnicas de aprendizaje supervisado con un balanceo adecuado de clases en escenarios de violencia de género, donde la minoría de casos más críticos exige especial atención. El paso siguiente consiste en afinar la integración de los modelos en herramientas de apoyo a la toma de decisiones, orientadas a priorizar la atención y los recursos en casos de mayor vulnerabilidad.

## **Discusiones**
En este proyecto, por razones de confidencialidad asociadas a la base de datos SIVIGILA, únicamente se empleó la información contenida en la primera hoja del formulario original. Si bien los modelos alcanzaron un desempeño aceptable con
las variables disponibles, la persistencia de un tamaño muy reducido de la clase minoritaria sugiere la necesidad de enriquecer el conjunto de datos para mejorar la discriminación de víctimas en riesgo. En particular, la incorporación de las variables de la segunda hoja del formulario—que incluyen datos del agresor (edad, relación con la víctima, antecedentes), modalidad de violencia (física, sexual, psicológica), lugar del suceso, tipo y gravedad
de las lesiones, tiempo de intervención, entre otros—podría aportar información crítica para elevar el F1-score macro. Estos atributos adicionales permiten:

* Caracterizar con mayor detalle el contexto del incidente.
* Identificar patrones de riesgo asociados a perfiles específicos de agresores.
* Discriminar mejor entre casos extremos (muerte) y hospitalización leve.
* Enriquecer las señales para los algoritmos de aprendizaje
supervisado.

Por tanto, una vez dispongamos de la segunda hoja de datos bajo los protocolos de confidencialidad, se recomienda rehacer el preprocesamiento y volver a entrenar los modelos—especialmente XGBoost y Random Forest—para evaluar la ganancia en métricas de interés.
